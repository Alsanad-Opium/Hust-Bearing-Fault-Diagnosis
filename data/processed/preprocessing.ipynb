```python
datapath = Path("D:/Coding/ML_projects/Hust-Bearing-Fault-Diagnosis/data/raw/HUST bearing dataset")
mat_files = list(datapath.rglob(pattern="*.mat"))
print("Total mat files present ", len(mat_files))
```

```python
attributes= []

for file  in mat_files:
    
    mat_at = loadmat(file,spmatrix=False, simplify_cells = True)
    
    attribute = {
        "fileName" : file.stem ,
        "attributes" : mat_at.keys()
    }
    attributes.append(attribute)
    
for attri in attributes:
    print(attri)
```

```python
records = []

for file in mat_files:
    
    try:
        
        matcontent = loadmat(file, spmatrix= False, simplify_cells =True) # load the mat file and converts the (N,1) or (1,N) (N) reduces the unecessay dimensional that the matlab adds
        
        record = {
            "recording_id": file.stem,
            "data": matcontent.get("data"),
            "fs":  matcontent.get("fs"),
            "rpm": matcontent.get("rpm") if matcontent.get("rpm") else None ,
            "ru": matcontent.get("ru") if matcontent.get("ru") else None,
            "ru_raw":matcontent.get("ru_raw") if matcontent.get("ru_raw") else None
            
        }
        records.append(record)
    except Exception as e:
        print(file , e)
```

```python
df = pd.DataFrame(records)
df.head()
```
