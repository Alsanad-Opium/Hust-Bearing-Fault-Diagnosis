```python
datapath = Path("D:/Coding/ML_projects/Hust-Bearing-Fault-Diagnosis/data/raw/HUST bearing dataset")
mat_files = list(datapath.rglob(pattern="*.mat"))
print("Total mat files present ", len(mat_files))
```

```python
records = []

for file in mat_files:
    
    try:
        
        matcontent = loadmat(file, spmatrix= False, simplify_cells =True) # load the mat file and converts the (N,1) or (1,N) (N) reduces the unecessay dimensional that the matlab adds
        
        record = {
            "recording_id": file.stem,
            "data": matcontent["data"],
            "fs":  matcontent["fs"],
            "rpm": matcontent["rpm"],
            "ru": matcontent["ru"]
            
        }
        records.append(record)
    except Exception as e:
        print(file , e)
```

```python
df = pd.DataFrame(records)
df.head()
```
